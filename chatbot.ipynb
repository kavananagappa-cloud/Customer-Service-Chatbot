```python

# ==========================================================
# PROJECT: CUSTOMER SERVICE CHATBOT USING NLP
# PLATFORM: GOOGLE COLAB
# LANGUAGE: PYTHON
# ==========================================================

# STEP 1: INSTALL AND IMPORT LIBRARIES

!pip -q install scikit-learn

import re
from datetime import datetime
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("=" * 55)
print("       CUSTOMER SERVICE CHATBOT")
print("          NLP BASED PROJECT")
print("=" * 55)


# STEP 2: CREATE CUSTOMER SERVICE DATASET

knowledge_base = {

    "greeting": {
        "examples": [
            "hello", "hi", "hey", "good morning",
            "good evening", "how are you"
        ],
        "response": "Hello! Welcome to ABC Store. How can I help you?"
    },

    "goodbye": {
        "examples": [
            "bye", "goodbye", "see you",
            "have a nice day"
        ],
        "response": "Thank you for contacting ABC Store. Have a great day!"
    },

    "thanks": {
        "examples": [
            "thank you", "thanks",
            "that was helpful", "thank you so much"
        ],
        "response": "You're welcome! Happy to help you."
    },

    "products": {
        "examples": [
            "what products do you sell",
            "tell me about your products",
            "show me your products",
            "product information",
            "what items are available"
        ],
        "response": "We offer different products including clothing, electronics and accessories."
    },

    "delivery": {
        "examples": [
            "how long does delivery take",
            "when will my order arrive",
            "delivery time",
            "how many days for shipping",
            "where is my parcel"
        ],
        "response": "Our standard delivery usually takes 3 to 5 business days in this demo."
    },

    "order_status": {
        "examples": [
            "track my order",
            "where is my order",
            "check order status",
            "my package has not arrived",
            "how can I track my parcel"
        ],
        "response": "Please check the tracking link in your order confirmation. This demo cannot access live order details."
    },

    "returns": {
        "examples": [
            "how can I return a product",
            "I want to return my order",
            "return policy",
            "can I exchange an item",
            "how do I replace a product"
        ],
        "response": "Eligible products can be returned within 7 days in this demo. Please contact customer support to start a return."
    },

    "refund": {
        "examples": [
            "when will I get my refund",
            "refund status",
            "my money has not come back",
            "how long does a refund take",
            "I want my money back"
        ],
        "response": "A demo refund may take 5 to 7 business days after return approval."
    },

    "payment": {
        "examples": [
            "which payment methods are accepted",
            "can I pay by upi",
            "do you accept cards",
            "payment options",
            "payment failed"
        ],
        "response": "Our sample store supports UPI, debit cards, credit cards and cash on delivery. No real payment is processed."
    },

    "timings": {
        "examples": [
            "what are your opening hours",
            "when does the store open",
            "when do you close",
            "store timings",
            "are you open today"
        ],
        "response": "Our sample store is open Monday to Saturday, from 9 AM to 8 PM."
    },

    "contact": {
        "examples": [
            "how can I contact customer care",
            "customer support phone number",
            "email address",
            "contact details",
            "talk to an agent"
        ],
        "response": "You can contact our demo support team at support@abcstore.example. Replace this with your approved contact details."
    },

    "location": {
        "examples": [
            "where is your store located",
            "store address",
            "what is your location",
            "where are you located"
        ],
        "response": "ABC Store is an imaginary store used for this project. Add your chosen store's address here."
    },

    "cancellation": {
        "examples": [
            "how to cancel my order",
            "I want to cancel my order",
            "cancel my product",
            "can I cancel an order"
        ],
        "response": "To cancel an order in a real store, contact customer support as soon as possible."
    }
}


# STEP 3: PREPARE DATA FOR NLP

training_sentences = []
training_intents = []

for intent, details in knowledge_base.items():

    for example in details["examples"]:

        training_sentences.append(example)
        training_intents.append(intent)


# STEP 4: APPLY TF-IDF

vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2)
)

training_vectors = vectorizer.fit_transform(training_sentences)

print("\nNLP Model Trained Successfully!")


# STEP 5: CREATE CHATBOT FUNCTION

def chatbot_reply(user_message):

    user_message = re.sub(
        r"[^a-zA-Z0-9\s]",
        " ",
        user_message.lower()
    ).strip()

    if not user_message:
        return "Please enter a valid question."

    # Convert user message into numerical form
    user_vector = vectorizer.transform([user_message])

    # Calculate similarity
    similarity_scores = cosine_similarity(
        user_vector,
        training_vectors
    )[0]

    # Find the closest matching question
    best_index = similarity_scores.argmax()

    best_score = similarity_scores[best_index]

    best_intent = training_intents[best_index]

    # Confidence threshold
    if best_score < 0.20:

        return (
            "Sorry! I couldn't understand your question. "
            "Please ask in another way or contact customer support."
        )

    # Return matching answer
    return knowledge_base[best_intent]["response"]


# STEP 6: DISPLAY CHATBOT INFORMATION

print("\nWelcome to ABC Store Customer Support!")
print("I am your virtual assistant.")
print("You can ask me about:")
print("- Products")
print("- Delivery")
print("- Order Status")
print("- Returns and Refunds")
print("- Payment")
print("- Store Timings")
print("- Contact Information")
print("- Order Cancellation")

print("\nType 'exit' to close the chatbot.")

print("\n" + "=" * 55)


# STEP 7: START CHATBOT CONVERSATION

while True:

    user_input = input("\nYou: ")

    if user_input.lower().strip() in ["exit", "quit", "stop"]:

        print("\nBot: Thank you for contacting ABC Store!")
        print("Have a nice day!")
        break

    response = chatbot_reply(user_input)

    print("\nChatbot:", response)


# ==========================================================
#             PROJECT EXECUTION COMPLETED
# ==========================================================
```
